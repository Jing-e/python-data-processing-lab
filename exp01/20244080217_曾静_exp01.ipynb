## 实验2：函数、条件、循环与递归

练习自定义函数、条件分支、`for`/`while` 循环以及递归/迭代两种实现方式。


```python
# 实验2-1：score_level(score)
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError(f"分数 {score} 超出 0~100 范围")
    if 90 <= score <= 100:
        return "优秀"
    if 80 <= score < 90:
        return "良好"
    if 60 <= score < 80:
        return "及格"
    return "不及格"

for s in [100, 95, 85, 80, 79, 60, 59, 0]:
    print(f"{s} -> {score_level(s)}")

try:
    score_level(101)
except ValueError as e:
    print(f"捕获异常: {e}")
```

```python
# 实验2-2：分别用 for 和 while 计算 1~100 的整数和
s_for = 0
for i in range(1, 101):
    s_for += i

s_while = 0
i = 1
while i <= 100:
    s_while += i
    i += 1

print("for 求和:  ", s_for)
print("while 求和:", s_while)
print("两者相等:", s_for == s_while)
```

```python
# 实验2-3：阶乘的递归版与迭代版（0! = 1，输入为非负整数）
def factorial_rec(n):
    if n < 0:
        raise ValueError("输入必须为非负整数")
    if n == 0:
        return 1
    return n * factorial_rec(n - 1)

def factorial_iter(n):
    if n < 0:
        raise ValueError("输入必须为非负整数")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

for n in [0, 1, 5, 10]:
    print(f"n={n}: 递归={factorial_rec(n)}, 迭代={factorial_iter(n)}")
```

```python
# 实验2-4：summary(values)，以元组返回最小值、最大值、平均值
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    return min(values), max(values), sum(values) / len(values)

data = [85, 58, 92, 76, 60, 89]
mn, mx, mean = summary(data)   # 元组解包
print(f"最小值={mn}, 最大值={mx}, 平均值={mean:.2f}")
```

```python
# 实验2-5：sum_all(*args)，返回所有参数之和
def sum_all(*args):
    return sum(args)

print("无参数:  ", sum_all())
print("单个参数:", sum_all(5))
print("多个参数:", sum_all(1, 2, 3, 4, 5))
```

## 实验3：列表、字典、元组与基础统计

使用给定模拟成绩数据，练习 `zip`、`enumerate`、列表推导式、统计与对象引用。


```python
# 实验3 数据
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]
```

```python
# 实验3-1：zip() 构建字典 + items() 逐项输出
score_dict = dict(zip(names, scores))
for name, score in score_dict.items():
    print(f"{name}: {score}")
```

```python
# 实验3-3：列表推导式
passed = [name for name, score in zip(names, scores) if score >= 60]  # 及格学生
squares = [s ** 2 for s in scores]                                     # 成绩平方
print("及格学生:", passed)
print("成绩平方:", squares)
```

## 实验4：文件读写、异常处理与模块

在 Notebook 同级目录创建 `exp01_data`，写入带无效记录的 `students.txt`，逐行解析、
捕获异常，并用 `pickle` 和自定义模块 `student_utils.py` 完成读写。


```python
# 实验4-1：查看工作目录、构造路径、创建数据目录
import os

cwd = os.getcwd()
data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)

print("当前工作目录:", cwd)
print("数据目录:    ", data_dir)
print("目录存在:    ", os.path.isdir(data_dir))
```

```python
# 实验4-2：写入 students.txt，再读取并跳过表头
data_content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

file_path = os.path.join(data_dir, "students.txt")
with open(file_path, "w", encoding="utf-8") as f:
    f.write(data_content)

with open(file_path, "r", encoding="utf-8") as f:
    lines = f.read().splitlines()

header = lines[0]
records = lines[1:]   # 跳过表头
print("表头:", header)
for line in records:
    print(repr(line))
```

```python
# 实验4-3：parse_student(line)，无效记录抛 ValueError
def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误: {len(parts)}")
    name, score_str = parts
    if not name.strip():
        raise ValueError("姓名为空")
    score = int(score_str)   # 非整数会抛 ValueError
    if not 0 <= score <= 100:
        raise ValueError(f"成绩 {score} 超出 0~100 范围")
    return {"name": name.strip(), "score": score}
```

```python
# 实验4-4：按行处理，捕获异常，有效记录存为字典列表
valid_records = []
invalid_records = []

for lineno, line in enumerate(records, start=1):
    try:
        valid_records.append(parse_student(line))
    except ValueError as e:
        invalid_records.append((lineno + 1, line, str(e)))  # +1 因表头占一行

print("有效记录:")
for r in valid_records:
    print(" ", r)
print("\n无效记录:")
for ln, line, reason in invalid_records:
    print(f"  第{ln}行: {line!r} -> {reason}")
```

```python
# 实验4-5：写报告，演示 w 与 a 模式
valid_count = len(valid_records)
invalid_count = len(invalid_records)
avg_score = sum(r["score"] for r in valid_records) / valid_count
pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / valid_count

report_path = os.path.join(data_dir, "report.txt")

# w 模式：覆盖写入（清空原有内容）
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {valid_count}\n")
    f.write(f"无效记录数: {invalid_count}\n")
    f.write(f"平均分: {avg_score:.2f}\n")
    f.write(f"及格率: {pass_rate:.1%}\n")

print("=== w 模式写入后 ===")
with open(report_path, encoding="utf-8") as f:
    print(f.read())

# a 模式：追加写入（保留原有内容，在末尾添加）
with open(report_path, "a", encoding="utf-8") as f:
    f.write("---- 追加内容：第二次写入 ----\n")

print("=== a 模式追加后 ===")
with open(report_path, encoding="utf-8") as f:
    print(f.read())
```

```python
# 实验4-6：pickle 序列化与反序列化
import pickle

pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

with open(pkl_path, "rb") as f:
    loaded = pickle.load(f)

print("恢复的数据与原数据相等:", loaded == valid_records)
```

## 实验5：类、继承与对象复制

定义 `Student` 类与继承类 `GraduateStudent`，并比较直接赋值、浅拷贝、深拷贝的区别。


```python
# 实验5-1：定义 Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name={self.name}, score={self.score}, 状态={status})"
```

```python
# 实验5-2：为实验4的每条有效记录创建 Student 对象
students = [Student(r["name"], r["score"]) for r in valid_records]
for stu in students:
    print(stu)
```

```python
# 实验5-3：GraduateStudent 继承 Student
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return super().__str__() + f", 研究方向={self.research_topic}"

g = GraduateStudent("Alice", 85, "数据分析")
print(g)
print("是否为 Student 的子类:", issubclass(GraduateStudent, Student))
```

```python
# 实验5-4：直接赋值、copy.copy、copy.deepcopy 对比
import copy

def make_obj():
    return {"name": "Alice", "scores": [85, 90]}

# 1. 直接赋值：同一个对象
o1 = make_obj()
ref = o1
ref["scores"][0] = 0
print("直接赋值后 原对象:", o1)      # 被改变

# 2. copy.copy：浅拷贝，内部列表仍共享
o2 = make_obj()
shallow = copy.copy(o2)
shallow["scores"][0] = 0
print("浅拷贝后   原对象:", o2)      # 内部列表被改变

# 3. copy.deepcopy：深拷贝，完全独立
o3 = make_obj()
deep = copy.deepcopy(o3)
deep["scores"][0] = 0
print("深拷贝后   原对象:", o3)      # 不变
```
