## 字符串的乐趣

处理字符串的三个简单函数：回文判断、字母序判断、去元音。本部分无需错误检查。

```python
# 1. is_palindrome：忽略空格与大小写判断回文
def is_palindrome(s):
    t = s.replace(" ", "").lower()   # 去掉空格并统一小写
    return t == t[::-1]

for s in ["level", "kayak", "pop", "tacocat", "T A C O cat",
          "rats live on no evil star", "hello"]:
    print(f"is_palindrome({s!r}) = {is_palindrome(s)}")
```

```python
# 2. is_abecedarian：忽略空格，判断字母是否按字母序排列
def is_abecedarian(s):
    t = s.replace(" ", "").lower()
    return t == "".join(sorted(t))

for s in ["adder", "beet", "dog", "cat", "abcd efgh xyz", "abc"]:
    print(f"is_abecedarian({s!r}) = {is_abecedarian(s)}")
```

```python
# 3. remove_vowels：去掉元音 a e i o u（不区分大小写）
def remove_vowels(s):
    return "".join(ch for ch in s if ch.lower() not in "aeiou")

for s in ["cat", "audio", "aeiou", "Hello World", "bcdfg"]:
    print(f"remove_vowels({s!r}) = {remove_vowels(s)!r}")
```

## 列表的乐趣

两个列表操作：反转列表、判断升序。

```python
# 1. list_reverse：返回反转后的列表，输入非列表抛出 TypeError
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]

print(list_reverse([1, 2, 3]))
print(list_reverse(["a", "b", "c"]))

try:
    list_reverse((1, 2, 3))
except TypeError as e:
    print("捕获 TypeError:", e)
```

```python
# 2. is_sorted：判断序列是否升序，只遍历一次
def is_sorted(seq):
    # 惰性生成器：遇到逆序立即短路，整体只遍历一遍
    return all(seq[i] <= seq[i + 1] for i in range(len(seq) - 1))

for s in [[1, 2, 3, 4], [1, 3, 2], (1, 2, 2, 5), "abc", "acb", []]:
    print(f"is_sorted({s!r}) = {is_sorted(s)}")
```

## 作为向量的元组

用元组表示向量与矩阵，实现标量乘、内积、矩阵校验与矩阵乘向量。

```python
# 1. vec_scalar_mult：标量乘，类型不合规抛 TypeError
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量必须是整数或浮点数")
    if not all(isinstance(x, (int, float)) for x in t):
        raise TypeError("元组元素必须是整数或浮点数")
    return tuple(x * s for x in t)

print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((1.5, 2.0), 3))

for bad in [([1, 2, 3], 2), ((1, 2, 3), "2"), ((1, "a", 3), 2)]:
    try:
        vec_scalar_mult(*bad)
    except TypeError as e:
        print("捕获 TypeError:", e)
```

```python
# 2. vec_inner_product：内积，非元组抛 TypeError，长度不一致抛 ValueError，返回浮点数
def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量长度不一致")
    return float(sum(a * b for a, b in zip(t1, t2)))

print(vec_inner_product((1, 2, 3), (4, 5, 6)))   # 1*4+2*5+3*6 = 32.0

try:
    vec_inner_product([1, 2], (3, 4))
except TypeError as e:
    print("捕获 TypeError:", e)

try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as e:
    print("捕获 ValueError:", e)
```

```python
# 3. check_valid_mx：判断是否为有效矩阵（元组的元组，各行等长，元素均为 int/float）
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if not all(isinstance(row, tuple) for row in mx):
        return False
    if len({len(row) for row in mx}) > 1:   # 各行长度不一致
        return False
    return all(isinstance(x, (int, float)) for row in mx for x in row)

print(check_valid_mx(((1, 2), (3, 4))))        # True
print(check_valid_mx(((1, 2), (3, 4, 5))))     # False（长度不一致）
print(check_valid_mx(((1, "a"), (3, 4))))      # False（含非数值）
print(check_valid_mx(((1, 2), [3, 4])))        # False（内层非元组）
print(check_valid_mx([(1, 2), (3, 4)]))        # False（外层非元组）
```

```python
# 4. mx_vec_mult：矩阵乘向量，输入不合法或维度不匹配抛异常
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise ValueError("矩阵不合法")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    if not all(isinstance(x, (int, float)) for x in vec):
        raise TypeError("向量元素必须是整数或浮点数")
    n = len(mx[0]) if mx else 0
    if len(vec) != n:
        raise ValueError("向量维度与矩阵列数不匹配")
    return tuple(sum(row[j] * vec[j] for j in range(n)) for row in mx)

mx = ((1, 2, 3), (4, 5, 6))   # 2x3
vec = (1, 1, 1)
print(mx_vec_mult(mx, vec))    # 每行求和：(6, 15)

try:
    mx_vec_mult(((1, 2), (3, 4, 5)), (1, 2))
except ValueError as e:
    print("捕获 ValueError:", e)
```

## 更多向量的乐趣：Vector 类

用面向对象方式实现向量，包含构造、比较、内积与乘法运算符。

```python
# 1. Vector 类：构造、比较、内积与乘法
class Vector:
    def __init__(self, dim, elems=None):
        if isinstance(dim, bool) or not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        if elems is None:
            elems = [0] * dim            # 只提供维度时默认零向量
        else:
            if not isinstance(elems, (list, tuple)):
                raise TypeError("元素必须是列表或元组")
            if len(elems) != dim:
                raise ValueError("维度与元素个数不一致")
        self.dim = dim                   # 实例属性
        self.elems = list(elems)         # 实例属性

    def __repr__(self):
        return f"Vector({self.dim}, {tuple(self.elems)})"

    # 3. 比较运算符：按字典序，维度不同抛异常
    def _check_comparable(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与另一个 Vector 比较")
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能比较")

    def __eq__(self, other):
        self._check_comparable(other)
        return tuple(self.elems) == tuple(other.elems)

    def __lt__(self, other):
        self._check_comparable(other)
        return tuple(self.elems) < tuple(other.elems)

    def __le__(self, other):
        self._check_comparable(other)
        return tuple(self.elems) <= tuple(other.elems)

    def __gt__(self, other):
        self._check_comparable(other)
        return tuple(self.elems) > tuple(other.elems)

    def __ge__(self, other):
        self._check_comparable(other)
        return tuple(self.elems) >= tuple(other.elems)

    # 4. dot：内积
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector")
        if self.dim != other.dim:
            raise ValueError("维度不一致")
        return sum(a * b for a, b in zip(self.elems, other.elems))

    # 5. 乘法：标量左乘/右乘、向量逐元素乘
    def __mul__(self, other):
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度不一致")
            return Vector(self.dim, tuple(a * b for a, b in zip(self.elems, other.elems)))
        if isinstance(other, (int, float)):
            return Vector(self.dim, tuple(x * other for x in self.elems))
        raise TypeError("只支持与标量或 Vector 相乘")

    def __rmul__(self, other):
        if isinstance(other, (int, float)):
            return Vector(self.dim, tuple(x * other for x in self.elems))
        raise TypeError("只支持标量左乘")
```

**关于类属性与实例属性的选择：** `dim` 与 `elems` 应作为**实例属性**。
因为每个 `Vector` 对象都有自己独立的维度和元素；若设为类属性，所有实例会共享同一个
列表，任何一个向量修改元素都会"传染"给其它所有向量，这与向量的语义不符。
因此用 `self.dim`、`self.elems` 在 `__init__` 中为每个对象分别赋值。

```python
# Vector 类的测试
v1 = Vector(2, (2, 4))
v2 = Vector(2, (2, 5))
v3 = Vector(2, (2, 4))
zero = Vector(3)          # 只提供维度 -> 默认零向量 (0, 0, 0)

print("v1 =", v1, " v2 =", v2, " v3 =", v3, " zero =", zero)
print("默认零向量 zero.elems =", zero.elems)

# 3. 比较运算符（字典序）
print("相等:", v1 == v3, " 不等:", v1 != v2)
print("字典序: v1<v2", v1 < v2, " v1<=v2", v1 <= v2, " v1>v2", v1 > v2, " v1>=v2", v1 >= v2)

try:
    _ = v1 == Vector(3, (1, 2, 3))
except ValueError as e:
    print("维度不同比较抛异常:", e)

# 4. dot
print("内积 v1.dot(v2) =", v1.dot(v2))   # 2*2 + 4*5 = 24

# 5. 乘法
a = Vector(2, (2, 3))
b = Vector(2, (4, 5))
print("a * b =", a * b)   # (8, 15)
print("2 * a =", 2 * a)   # (4, 6)
print("a * 2 =", a * 2)   # (4, 6)

try:
    _ = a * Vector(3, (1, 2, 3))
except ValueError as e:
    print("维度不一致相乘抛异常:", e)

# 构造异常测试
for args in [(-1, (1, 2)), (2, (1, 2, 3)), ("2", (1, 2)), (2, "ab")]:
    try:
        Vector(*args)
    except (TypeError, ValueError) as e:
        print(f"Vector{args} 抛异常: {type(e).__name__}: {e}")
```
