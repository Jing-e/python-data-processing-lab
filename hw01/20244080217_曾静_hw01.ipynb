## 欧几里得算法（GCD）

欧几里得算法用于求两个整数的最大公约数（GCD）。核心思想：`gcd(m, n) = gcd(n, m % n)`，
反复取余直到余数为 0，此时另一个数即为 GCD。


```python
# 实现 gcd 函数
def gcd(m, n):
    while n:
        m, n = n, m % n
    return m

# 计算三组数值的 GCD
print("gcd(2018, 2019) =", gcd(2018, 2019))
print("gcd(1200, 300)  =", gcd(1200, 300))
print("gcd(5040, 60)   =", gcd(5040, 60))
```

### 负数输入的行为讨论

Python 的取余运算 `%` 在遇到负数时，结果的符号与除数一致，因此上述 `gcd` 实现
在输入为负数时可能返回**负数**的公约数。数学上 GCD 定义为非负整数，所以这种
行为略有瑕疵；可以在返回前加 `abs()`，或直接使用 `math.gcd()`（它保证返回非负）。


```python
# 观察负数输入的行为
print("gcd(2018, -2019) =", gcd(2018, -2019))
print("gcd(-2018, 2019) =", gcd(-2018, 2019))
print("gcd(-2018, -2019) =", gcd(-2018, -2019))

# 修正：返回非负 GCD
def gcd_fixed(m, n):
    return abs(gcd(m, n))

print("gcd_fixed(2018, -2019) =", gcd_fixed(2018, -2019))
```

## 近似欧拉数 e

自然对数的底 e 可用两种方式近似：

1. **极限形式**（伯努利）：$e = \lim_{n \to \infty}(1 + \frac{1}{n})^n$
2. **无穷级数**：$e = \sum_{k=0}^{\infty}\frac{1}{k!} = 1 + \frac{1}{1!} + \frac{1}{2!} + \cdots$

下面分别实现相关函数。


```python
import math

# euler_limit(n)：用 (1 + 1/n)^n 近似 e
def euler_limit(n):
    return (1 + 1 / n) ** n

print("euler_limit(1)    =", euler_limit(1))
print("euler_limit(10)   =", euler_limit(10))
print("euler_limit(1000) =", euler_limit(1000))
print("真实 e            =", math.exp(1))
```

```python
# euler_infinite_sum(n)：前 n 项和（1-indexed）
#   即 k = 0..n-1 的 1/k! 之和；约定 0! = 1
def euler_infinite_sum(n):
    total = 0.0
    for k in range(n):
        total += 1.0 / math.factorial(k)
    return total

print("euler_infinite_sum(0)  =", euler_infinite_sum(0))   # 0
print("euler_infinite_sum(1)  =", euler_infinite_sum(1))   # 1/0! = 1
print("euler_infinite_sum(4)  =", euler_infinite_sum(4))   # 1 + 1 + 1/2 + 1/6
print("euler_infinite_sum(10) =", euler_infinite_sum(10))
```

```python
# euler_approx(epsilon)：用级数逼近 e，直到误差 < epsilon
def euler_approx(epsilon):
    true_e = math.exp(1)
    approx = 0.0
    k = 0
    while abs(approx - true_e) >= epsilon:
        approx += 1.0 / math.factorial(k)
        k += 1
    return approx

print("euler_approx(0.001) =", euler_approx(0.001))
print("euler_approx(1e-6)  =", euler_approx(1e-6))
print("真实 e              =", math.exp(1))
```

```python
# print_euler_sum_table(n)：打印 k=1..n 的 euler_infinite_sum(k)，每行一个
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(f"euler_infinite_sum({k}) = {euler_infinite_sum(k)}")

print_euler_sum_table(10)
```

## 测试整数属性

判断一个正整数是否为 2 的幂（$n = 2^p$），并推广到判断是否为任意底数 b 的幂。
要求：**不能使用 `math.sqrt`**，只用除法（`//`）和取模（`%`）运算。


```python
# is_power_of_2(n)：n 是否为 2 的幂
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

for n in [1, 2, 4, 8, 16, 32, 6, 10, 100]:
    print(f"is_power_of_2({n}) = {is_power_of_2(n)}")
```

```python
# is_power(b, n)：n 是否为 b 的幂
def is_power(b, n):
    if n == 1:
        return True        # 1 = b^0
    if b == 1:
        return False       # 1 的任意次幂都是 1，n>1 不可能是 1 的幂
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

print("is_power(2, 8)   =", is_power(2, 8))    # True: 8 = 2^3
print("is_power(3, 9)   =", is_power(3, 9))    # True: 9 = 3^2
print("is_power(5, 125) =", is_power(5, 125))  # True: 125 = 5^3
print("is_power(2, 10)  =", is_power(2, 10))   # False
print("is_power(2, 1)   =", is_power(2, 1))    # True: 1 = 2^0
print("is_power(3, 27)  =", is_power(3, 27))   # True: 27 = 3^3
```
